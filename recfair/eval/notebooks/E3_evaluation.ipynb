# B. Hipótese

O workflow E2 falha de três modos **medidos**: claims por substring (T18/T21/T22), guardrail de mentira (T26–T30) e domínio único (sem FAQ/revenda). A hipótese E3 é que um **supervisor** + matcher semântico + sanitização regex + RAG FAQ fecha esses gaps sem reescrever o engine de 7 passos.

Não prometemos ganho em T09/T17/T20/T31 (intent/memória). Empate ou queda em **H.1** (recomendação) com ganho em **H.2** (segurança) ou **H.3** (FAQ) é resultado válido (seção J).

**Por que exact-match sozinho é insuficiente:** T12 (mesmos 5 SKUs, desempate trocado) deveria ser near-miss; T14 (categoria/janela errada) é erro grave. nDCG@5 ≈ 1,0 no primeiro e baixo no segundo (Valcarce et al. 2020; Bauer et al. TORS 2024; [Evidently](https://www.evidentlyai.com/ranking-metrics/evaluating-recommender-systems)). `gap_intentional_pass` exige a lista **filtrada** e rejeita coincidência com a lista ingênua E1 — senão um filtro futuro que muda o gabarito continuaria “passando” por popularidade.

````python
gold = ["A", "B", "C", "D", "E"]
print("T12-sim nDCG@5 permutação", round(ndcg_at_5(["B", "A", "C", "D", "E"], gold), 4))
print("T14-sim nDCG@5 categoria errada", round(ndcg_at_5(["X", "Y", "Z", "W", "V"], gold), 4))
````

# I. Novos modos de falha

Introduzidos pelo multiagente (a relatar no run, quando houver):

1. Supervisor encaminha ao especialista errado (H.4 — T41–T43 e T52–T58).
2. Especialista fora do escopo (FAQ com SKUs; recomendação em política).
3. Contexto corrompido (query sanitizada demais; chunks FAQ ignorados).
4. Estouro de `recursion_limit` (só há 1 replan; estouro ainda é falha).
5. Tool certa com argumento errado (`retrieve_faq` / `match_claims_semantic`).
6. Tool necessária não chamada, ou tool desnecessária.
7. Erro de tool silencioso (proibido pelo harness — deve aparecer em `AgentResult.status=error`).
8. FAQ inventa além da evidência.
9. Guardrail falso positivo (redact quebra intent) ou falso negativo (PII vaza).

Da régua (não são falhas do agente): T12 vs T14; `false_positive_gap`; tightening de `G_need` pode derrubar `e1_rate_overall` legado — o E3 lê `gap_intentional_pass`.

nDCG no **runtime** (ranking online) fica como evolução futura da régua, sem implementação aqui.

# J. Análise e pergunta obrigatória

**O que o E2 não resolvia e o E3 ataca:** domínio único, claims substring, PII/injection sem sanitização.

**O que permanece:** erros de intent/memória (T09, T31). Ranking continua no engine determinístico — o especialista de recomendação é um agente que **chama** um pipeline, não um LLM ranqueador.

**O que surgiu:** custo do supervisor, risco de rota errada, dependência MiniLM/FAISS.

**Poderia algum agente voltar a ser etapa?** Segurança e handoff já são etapas. FAQ sem evidência medida de domínio distinto voltaria a ser recusa genérica — o golden T39–T40 justifica o especialista.

### Qual versão levaria para uso real, e que evidência ainda falta?

**Para produção de catálogo (Top-5), o `workflow` E2 ainda é o núcleo de ranking** — determinístico, barato, auditável. **Para um assistente que também fala de política, pagamento e revenda, o `multiagent` é o único que tem rota e transbordo.** A escolha operacional depende do run H desta sessão: se **H.1** (nDCG@5) não cair além do intervalo esperado e **H.2**/**H.3**/**H.4** subirem, levaria o multiagent com o engine E2 encapsulado. Falta evidência: manifests `eval/runs/*.json` com os três `run_eval` no mesmo modelo; n maior que 43 casos; e um teste de falso positivo do guardrail em queries legítimas com CPF no texto da tarefa.

Ponto de partida do E4: se o Painel A cair por roteamento, especializar o supervisor (ou regras antes do LLM) antes de adicionar MCP ou um quarto agente.
