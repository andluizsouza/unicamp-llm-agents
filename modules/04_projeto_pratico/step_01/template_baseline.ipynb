# 15. Implementação da verificação

**[PREENCHER COM CÓDIGO]**

Implemente as verificações correspondentes aos critérios da seção 10.

As funções abaixo são um ponto de partida (adaptado do notebook da Aula 2); **adapte ou substitua**
conforme o problema do grupo. O que importa é que a verificação seja explícita e reproduzível.

```python
import re, unicodedata
from typing import Optional

def normalizar(texto: str) -> str:
    texto = unicodedata.normalize("NFKD", texto.lower())
    texto = "".join(c for c in texto if not unicodedata.combining(c))
    return re.sub(r"\s+", " ", texto).strip()

def cobertura_esperada(resposta: str, esperado: list) -> float:
    """Fração dos itens de referência presentes na resposta (correção + completude)."""
    texto = normalizar(resposta)
    return sum(normalizar(k) in texto for k in esperado) / len(esperado)

# Cuidado com verificações ingênuas: procurar a palavra "não" na resposta aprova
# uma saída que nega uma coisa e inventa outra. Exija coerência da saída inteira.
#
MARCADORES_AUSENCIA = [
    "nao esta", "nao consta", "nao foi encontrad", "nao informad",
    "nao especificad", "nao ha informacao", "nao menciona",
]

def admite_ausencia(resposta: str) -> bool:
    return any(m in normalizar(resposta) for m in MARCADORES_AUSENCIA)

print("[done]")
```
